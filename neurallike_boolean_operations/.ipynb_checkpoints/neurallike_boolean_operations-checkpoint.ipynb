# Neural-Like Boolean Operations

These examples are based on the introductory lessons of this Neps Academy course: <https://neps.academy/br/course/redes-neurais-com-pytorch>.
Even though the code is mine, the logic behind this implementation of the boolean operations is quite the same as in the lessons, so the idea is credited to them.

There is no real machine learning here. Just some of its concepts.

## Initial Definitions:

```python
class Neuron:
    def __init__(self, W: list[float], b: float, F = lambda x: x) -> None:
        self.W = W
        self.b = b
        self.F = F

    def get_n_inputs(self) -> int:
        return len(self.W)
        
    def get_W(self) -> list[float]:
        return self.W
        
    def set_W(self, W: list[float]) -> None:
        self.W = W
        
    def get_b(self) -> float:
        return self.b
        
    def set_b(self, b: float) -> None:
        self.b = b

    def score(self, X: list[float]) -> float:
        logit = sum([x * w for x, w in zip(X, self.W)]) + self.b
        score = self.F(logit)
        return score

    def __call__(self, X: list[float]) -> float:
        return self.score(X)
```

```python
class Layer:
    def __init__(self, neurons = list[Neuron]) -> None:
        self.neurons = neurons

    def get_n_neurons(self) -> int:
        return len(self.neurons)

    def get_n_inputs(self) -> int:
        return len(self.neurons[0].get_W())

    def score(self, X: list[float]) -> list[float]:
        scores = [neuron(X) for neuron in self.neurons]
        return scores
        
    def __call__(self, X: list[float]) -> list[float]:
        return self.score(X)
```

```python
class Network:
    def __init__(self, layers = list[Layer]) -> None:
        self.layers = layers

    def get_shape(self) -> list[int]:
        return [self.layers[0].get_n_inputs()] + [layer.get_n_neurons for layer in self.layers]

    def score(self, X: list[float]) -> list[float]:
        for layer in self.layers:
            X = layer.score(X)
        scores = X
        return scores
        
    def __call__(self, X: list[float]) -> list[float]:
        return self.score(X)
```

## Definitions For Boolean Cases

```python
cases = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]
```

```python
def test_boolean_op(nn: Network) -> None:
    for X in cases:
        print(X, nn(X))
```

## And

```python
and_op = Network([
    Layer([Neuron([1.0, 1.0], -1.5, lambda x: 1 if x > 0 else 0)])
])
```

```python
test_boolean_op(and_op)
```

## Or

```python
or_op = Network([
    Layer([Neuron([1.0, 1.0], -0.5, lambda x: 1 if x > 0 else 0)])
])
```

```python
test_boolean_op(or_op)
```

## XOr

```python
xor_op = Network([
    Layer([Neuron([1.0, 1.0], -1.5, lambda x: 1 if x > 0 else 0), Neuron([1.0, 1.0], -0.5, lambda x: 1 if x > 0 else 0)]),
    Layer([Neuron([-1.0, 1.0], -0.5, lambda x: 1 if x > 0 else 0)])
])
```

```python
test_boolean_op(xor_op)
```
